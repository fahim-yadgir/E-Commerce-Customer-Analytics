```python
def exctract_text(text):
    total = ""
    check = True
    for i in text:
        if i == ",":
            break
        if check:
            total+=i
    return total
        
```

```python
df["Specific_rating"] = df["review_text"].apply(exctract_text)
```

```python
def Text_inormation(text):
    check = False
    Total = ""
    for i in text:
        if i == ",":
            check = True
            continue 
        if i == ".":
            check = False
            break
        if check:
            Total+=i
    return Total
```

```python
df["Text_info"] = df["review_text"].apply(Text_inormation)
```
